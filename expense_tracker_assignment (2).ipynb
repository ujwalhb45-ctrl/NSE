```python
expenses = []  # list of dictionaries
filename = 'expenses.txt'
```

```python
def add_expense():
    try:
        amount = float(input('Enter amount: '))
        category = input('Enter category: ').strip().title()
        note = input('Enter note: ').strip()
        expense = {'amount': amount, 'category': category, 'note': note}
        expenses.append(expense)
        print('Expense added successfully')
    except ValueError:
        print('Invalid amount entered')
```

```python
def view_expenses():
    if not expenses:
        print('No expenses found')
    for i, exp in enumerate(expenses, start=1):
        print(i, exp.get('category'), exp.get('amount'), exp.get('note'))
```

```python
def search_expense():
    key = input('Enter category to search: ').strip().title()
    found = False
    for exp in expenses:
        if exp.get('category') == key:
            print(exp)
            found = True
    if not found:
        print('No matching expenses found')
```

```python
def save_to_file():
    with open(filename, 'w') as f:
        for exp in expenses:
            f.write(f"{exp['amount']},{exp['category']},{exp['note']}\n")
    print('Expenses saved to file')
```

```python
def load_from_file():
    try:
        with open(filename, 'r') as f:
            expenses.clear()
            for line in f:
                amt, cat, note = line.strip().split(',')
                expenses.append({'amount': float(amt), 'category': cat, 'note': note})
        print('Expenses loaded from file')
    except FileNotFoundError:
        print('No saved file found')
```

```python
def menu():
    while True:
        print('\n1.Add Expense\n2.View Expenses\n3.Search Expense\n4.Save\n5.Load\n6.Exit')
        choice = input('Enter choice: ')
        if choice == '1':
            add_expense()
        elif choice == '2':
            view_expenses()
        elif choice == '3':
            search_expense()
        elif choice == '4':
            save_to_file()
        elif choice == '5':
            load_from_file()
        elif choice == '6':
            print('Exiting program')
            break
        else:
            print('Invalid choice')
```
